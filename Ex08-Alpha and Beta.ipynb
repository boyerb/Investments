# Example Chapter 8: Estimating Alpha and Beta
In this example, we estimate the alphas and betas for decile portfolios posted in the Ken French Data Library.

### Imports and Setup

```python
# Install the current farms package and import the tools used below.
%pip install -q --upgrade farms

import farms as fm
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
pd.set_option('display.max_rows', 20)
```

### Load in Data Using the Python Function
See Ex07-Ken_French.ipynb for a complete describption of the function in the course GiHub repo.


```python
# Load monthly short-term-reversal decile returns and market factors.
df = fm.load_ken_french_data(
    "deciles",
    strategy="shorttermreversal",
    include_factors="market",
    details=True,
)

df.head()
```

### Estaimte Alpha and Beta

```python
# Estimate alpha and beta for decile 1 relative to the market portfolio.
df["dec1_excess"] = df["dec1"] - df["rf"]
results_dec1 = fm.linear_regression_summary(df["mkt-rf"], df["dec1_excess"])
results_dec1
```

```python
# Estimate alpha and beta for decile 10 relative to the market portfolio.
df["dec10_excess"] = df["dec10"] - df["rf"]
results_dec10 = fm.linear_regression_summary(df["mkt-rf"], df["dec10_excess"])
results_dec10
```
