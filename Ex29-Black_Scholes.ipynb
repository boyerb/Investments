# Example Chapter 29: Option Pricing with the Black-Scholes Model
In this example we use the `black_scholes()` function from the `farms` package to price a European call and put. The inputs match the Black-Scholes example in Chapter 29: a stock with a current price of \$100, a strike price of \$105, six months to expiration, an effective annual risk-free rate of 1%, and a volatility of 25%.

### Imports and Setup

```python
# Import packages
%pip install -q --upgrade farms

import farms as fm
```

### Set the Inputs
The risk-free rate $R_f$ is an effective annual rate. The Black-Scholes formula uses the continuously compounded rate $r_c = \ln(1+R_f)$, and `black_scholes()` makes this conversion for us.

```python
S = 100        # current stock price
K = 105        # strike price
T = 0.5        # time to expiration in years (six months)
rf = 0.01      # effective annual risk-free rate (1%)
sigma = 0.25   # annual volatility (25%)
```

### Price the Call and the Put
For full parameter details, see the [`farms` API documentation](https://github.com/boyerb/farms/blob/main/docs/api.md#black_scholes).

`black_scholes()`

**Required Inputs**

- `type` — Option type. Potential values are `"call"` and `"put"`. Example: `type='call'`.
- `S` — Current price of the underlying stock. Example: `S=100`.
- `K` — Strike price. Example: `K=105`.
- `T` — Time to expiration in years. Example: `T=0.5`.
- `rf` — Effective annual risk-free rate $R_f$, as a decimal. The function converts it to the continuously compounded rate $\ln(1+R_f)$. Example: `rf=0.01`.
- `sigma` — Annual volatility of the stock return, as a decimal. Example: `sigma=0.25`.

**Optional Inputs**

- None.

**Outputs**

- `float` — The Black-Scholes option price.

```python
call_price = fm.black_scholes('call', S, K, T, rf, sigma)
put_price = fm.black_scholes('put', S, K, T, rf, sigma)

print(f'Call price: {call_price:.4f}')
print(f'Put price:  {put_price:.4f}')
```

### Check Put-Call Parity
For European options on a stock that pays no dividends, put-call parity says

$$C - P = S - \frac{K}{(1+R_f)^T}.$$

Both sides should be equal.

```python
print(f'C - P:              {call_price - put_price:.4f}')
print(f'S - K/(1+R_f)^T:    {S - K / (1 + rf) ** T:.4f}')
```
