## Variables and lists

A **variable** is a name that stores a value. The equal sign (=) assigns the value on the right to the name on the left. A **list** is an ordered collection of values written inside square brackets. In the cell below, change one of the return values and run the cell again.

```python
# Store several possible returns in a list
returns = [-0.05, 0.00, 0.05, 0.10]

# Store the number of possible returns in a variable
number_of_possible_returns = len(returns)

print('Possible returns:', returns)
print('Number of possible returns:', number_of_possible_returns)
```

## Packages and functions

Python has many reusable collections of code called **packages** (also called libraries). A package contains many useful functions. We will use the NumPy package for numerical work. The statement `import numpy as np` makes the package available using the short name `np`.

A **function** takes inputs and produces an output. The `mean` function is one of the many functions provided by NumPy, so `np.mean(returns)` calculates the average of the values in `returns`.

```python
# Import the NumPy package and give it the standard short name, np
import numpy as np

# np.mean is a NumPy function that calculates an average
average_return = np.mean(returns)
print(f'Average of the listed returns: {average_return:.3f}')
```

## NumPy arrays

A NumPy **array** is a data structure for numerical values. It is similar to a list, but NumPy arrays make it easy to perform calculations on many values at once.

Suppose, for example, that you wanted to add 0.01 to every number in your list defined above. Run the next cell and examine the error message.

```python
# This does not add 0.01 to every item in a Python list
returns + 0.01
```

A Python list is mainly a container that holds several separate items. Python does not interpret `returns + 0.01` as “add 0.01 to every return,” so it raises a `TypeError`. A list can be added to another list, but Python interprets that as joining the lists.

A NumPy array is designed for numerical calculations. With a NumPy array, `+ 0.01` means “add 0.01 to each value in the array.” The next cell converts the list to an array and then adds 1 percentage point to every return without an error.

```python
returns_array = np.array(returns)
adjusted_returns = returns_array + 0.01

print('NumPy array:', returns_array)
print('Returns after adding 1 percentage point:', adjusted_returns)
```

## Objects and random samples

So far, our variables have stored simple values, such as a number or a list of returns. Variables can also store an **object**: a specialized tool that can hold information and perform particular actions. For example, a calculator is an object that can perform actions such as addition and multiplication.

In the next cell, we create a random-number-generator object: a tool that can generate random samples of returns. The expression `rng = np.random.default_rng()` creates this object and stores it in the variable `rng`. The name `rng` is short for random-number generator. You can think of `rng` as a random-number machine.

The expression `rng.choice(returns, size=10)` tells that particular random-number machine to choose 10 values from `returns`. The part after the period, `choice(...)`, is called a **method**: it is an action available on that particular object. Run the cell below more than once to see a different random sample.

```python
# Create a random-number-generator object
rng = np.random.default_rng()

# Tell the random-number generator to choose 10 returns at random
sampled_returns = rng.choice(returns, size=10)
print('Random sample:', sampled_returns)
print(f'Sample average: {np.mean(sampled_returns):.3f}')
```

## Key terms

- **Package (library):** reusable code that provides functions and classes.
- **Function:** reusable code that takes inputs and produces an output.
- **Method:** an action available on a particular object; for example, `rng.choice(...)`.
- **Object:** a particular tool stored in a variable; for example, `rng` is a random-number generator.
- **NumPy array:** a collection of numerical values designed for efficient computation.
