# Example Chapter 29: Implied Volatility with the Black-Scholes Model
Volatility is the only Black-Scholes input we cannot observe directly. The **implied volatility** is the value of $\sigma$ that makes the Black-Scholes price equal to the option's market price. In this example we use the `implied_volatility()` function from the `farms` package to find it.

### Imports and Setup

```python
# Import packages
%pip install -q --upgrade farms

import farms as fm
```

### Find the Implied Volatility from a Call Price
A call with a strike price of \$21 and three months to expiration trades for \$0.3942 when the stock price is \$20 and the effective annual risk-free rate is 1%. The function searches for the volatility that reproduces this price.

For full parameter details, see the [`farms` API documentation](https://github.com/boyerb/farms/blob/main/docs/api.md#implied_volatility).

`implied_volatility()`

**Required Inputs**

- `type` — Option type. Potential values are `"call"` and `"put"`. Example: `type='call'`.
- `option_price` — Observed market price of the option. Example: `option_price=0.3942`.
- `S` — Current price of the underlying stock. Example: `S=20`.
- `K` — Strike price. Example: `K=21`.
- `T` — Time to expiration in years. Example: `T=0.25`.
- `r` — Effective annual risk-free rate $R_f$, as a decimal. Example: `r=0.01`.

**Optional Inputs**

- None.

**Outputs**

- `float` — The implied volatility, searched between `1e-6` and `5.0`.
- `numpy.nan` — Returned when no volatility in that range matches the option price.

```python
call_price = 0.3942   # market price of the call
S = 20                # current stock price
K = 21                # strike price
T = 0.25              # time to expiration in years (three months)
rf = 0.01             # effective annual risk-free rate (1%)

implied_vol = fm.implied_volatility('call', call_price, S, K, T, rf)
print(f'Implied volatility: {implied_vol:.4f}')
```

### Check the Answer
If we plug the implied volatility back into `black_scholes()`, we should get the market price of the call.

For full parameter details, see the [`farms` API documentation](https://github.com/boyerb/farms/blob/main/docs/api.md#black_scholes).

`black_scholes()`

**Required Inputs**

- `type` — Option type. Potential values are `"call"` and `"put"`. Example: `type='call'`.
- `S` — Current price of the underlying stock. Example: `S=100`.
- `K` — Strike price. Example: `K=105`.
- `T` — Time to expiration in years. Example: `T=0.5`.
- `rf` — Effective annual risk-free rate $R_f$, as a decimal. The function converts it to the continuously compounded rate $\ln(1+R_f)$. Example: `rf=0.01`.
- `sigma` — Annual volatility of the stock return, as a decimal. Example: `sigma=0.25`.

**Optional Inputs**

- None.

**Outputs**

- `float` — The Black-Scholes option price.

```python
model_price = fm.black_scholes('call', S, K, T, rf, implied_vol)
print(f'Black-Scholes price at the implied volatility: {model_price:.4f}')
```
