#Example Chapter 8: Running Regressions Using Python
In this example we show how to load in data, run regressions using Python, and estimate the 95% confidence intervals for each parameter.

We install and import the current `farms` package. Its `linear_regression_summary()` function estimates the regression coefficients and confidence intervals for both single-variable and multiple regressions.

```python
# Install the current farms package and import the tools used below.
%pip install -q --upgrade farms

import farms as fm
import pandas as pd

pd.set_option('display.max_rows', 20)
```

### Load in Data
We now load a the data from the Excel Examples workbook under the `CH8 Los Angeles Lakers` tab into a pandas DataFrame. We explicitly choose a set of columns to import and specify the header row to ensure the data aligns correctly. Finally, we display the DataFrame contents.

```python
# Load the Lakers data from the Chapter 8 Excel examples workbook.
url = 'https://github.com/boyerb/Investments/raw/master/Examples_26.09.xlsx'
columns_to_read = ['Season', 'Leading Scorer', 'PPG', 'Pace', 'Win Percentage']
df = pd.read_excel(
    url,
    sheet_name='CH8 Los Angeles Lakers',
    header=3,
    usecols=columns_to_read,
    engine='openpyxl',
)
df
```

### Run Regression
Here we define our explanatory and dependent variables, run the regression, and then pull the output we want:
1. parameter estimates,
2. 95% confindence interval for each parameter

```python
X = df[["PPG", "Pace"]]
Y = df["Win Percentage"]

results = fm.linear_regression_summary(X, Y)
results
```
